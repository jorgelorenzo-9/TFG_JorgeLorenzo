```python
from utilities import *
import pandas as pd
import statsmodels.api as sm
from statsmodels.discrete.discrete_model import Logit
import matplotlib.pyplot as plt

# Abrir archivos originales (se usan para calcular x1 y x2)
df1 = pd.read_csv("series_concatenadas_rampa/archivo_concatenado_miercoles_diff_8se_atk.csv", header=None)
df2 = pd.read_csv("series_concatenadas_sin_ataque/archivo_concatenado_miercoles_sin_ataque_diff.csv", header=None)

# Ruta del archivo generado por MATLAB
file_path = "matlab_bub/alpha_fit_normalized_miercoles_concatenado_rampa_desl_p1_8se_atk.csv"
df = pd.read_csv(file_path, sep=",")
print("Columnas del DataFrame:", df.columns.tolist())

# Calcular x1 y x2 usando ventanas (se asume 898 muestras por ventana y se usan los segmentos 720 a 898)
x1 = pd.Series([
    (float(x_new.mean().iloc[0]), float(x_new.std().iloc[0]))
    for x_new in [
        (df1[898*j+720:898*(j+1)] - df1[898*j:898*j+720].mean()) / df1[898*j:898*j+720].std()
        for j in range(len(df1) // 898)
    ]
])
x2 = pd.Series([
    (float(x_new.mean().iloc[0]), float(x_new.std().iloc[0]))
    for x_new in [
        (df2[898*j+720:898*(j+1)] - df2[898*j:898*j+720].mean()) / df2[898*j:898*j+720].std()
        for j in range(len(df2) // 898)
    ]
])
# Alternar x1 y x2 concatenándolos
xMeanStd = pd.concat([x1, x2], axis=0)

# Ahora, en lugar de asignar directamente (lo que produce error por desajuste de longitudes),
# distribuimos uniformemente los 16 pares en las filas del DataFrame 'df'.
n_blocks = len(xMeanStd)    # Por ejemplo, 16
n_rows = len(df)            # Por ejemplo, 128
k = n_rows // n_blocks      # Número base de repeticiones
remainder = n_rows % n_blocks  # Sobrante

mean_values = []
std_values = []
for i, (mean_val, std_val) in enumerate(xMeanStd):
    reps = k + (1 if i < remainder else 0)
    mean_values.extend([mean_val] * reps)
    std_values.extend([std_val] * reps)

df["mean_values"] = mean_values
df["std_values"] = std_values

# Seleccionar solo las variables relevantes para el modelo (en este ejemplo se usan beta, gamma y std)
selected_features = ["beta_values", "gamma_values", "delta_values", "mean_values", "std_values"]
#selected_features = ["alpha_values", "beta_values", "gamma_values", "delta_values", "mean_values", "std_values"] #alpha values es el que da problemas, sin el funciona 
Y = df["attack_value"].astype(float)  # Variable dependiente
X = df[selected_features].astype(float)  # Variables predictoras

# Agregar una constante
X = sm.add_constant(X)

# Entrenar modelo de regresión logística
model = Logit(Y, X)
results = model.fit()
print(results.summary())

# REPRESENTACIÓN CON SVM
from sklearn import svm
clf = svm.SVC(class_weight="balanced")
clf.fit(X, Y)

# Predicciones y matriz de confusión
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
predictions = clf.predict(X)
cm = confusion_matrix(Y, predictions, normalize='true')
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=["No Attack", "Attack"])
disp.plot(cmap=plt.cm.Blues)
plt.show()

# Métricas
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
accuracy = accuracy_score(Y, predictions)
precision = precision_score(Y, predictions)
recall = recall_score(Y, predictions)
f1 = f1_score(Y, predictions)
print("Accuracy: ", accuracy)
print("Precision: ", precision)
print("Recall: ", recall)
print("F1 score: ", f1)

# Curva ROC
from sklearn.metrics import roc_curve, auc
fpr, tpr, _ = roc_curve(Y, predictions)
roc_auc = auc(fpr, tpr)
plt.figure()
lw = 2
plt.plot(fpr, tpr, color='darkorange', lw=lw, label='ROC curve (area = %0.2f)' % roc_auc)
plt.plot([0, 1], [0, 1], color='navy', lw=lw, linestyle='--')
plt.xlim([0.0, 1.0])
plt.ylim([0.0, 1.05])
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('Receiver Operating Characteristic, all parameters')
plt.legend(loc="lower right")
plt.show()
```
